```python
Orders=pd.read_excel("sample_superstore.xls",sheet_name="Orders")
People=pd.read_excel("sample_superstore.xls",sheet_name="People")
Returns=pd.read_excel("sample_superstore.xls",sheet_name="Returns")
```

## Verifying Dates

```python
Orders_clean[['Order Date','Ship Date']].head(3)
```

```python
Orders_clean['Order Year']=Orders_clean['Order Date'].dt.year
```

```python
Orders_clean['Order Month']=Orders_clean['Order Date'].dt.month
```

```python
Orders_clean['Month Name']=Orders_clean['Order Date'].dt.month_name()
```

```python
Orders_clean["Quarter"] = Orders_clean["Order Date"].dt.quarter
```

```python
Orders_clean["Day Name"] = Orders_clean["Order Date"].dt.day_name()
```

## query to find the no of days it takes to ship the order 

```python
Orders_clean['Shipping Time']=(Orders_clean['Ship Date']-Orders_clean['Order Date'])
```

## Profit Margin 

```python
Orders_clean['Profit Margin %']=Orders_clean['Profit']/Orders_clean['Sales']*100
```

## Discount %

```python
Orders_clean['Discount %']=Orders_clean['Discount']*100
```

```python
Orders_clean.head(2)
```

## Negative Sales

```python
Orders_clean[Orders_clean["Sales"] < 0]
```

## Negative Quantity 

```python
Orders_clean[Orders_clean["Quantity"] < 0]
```

## Checking for Discount more than 100 %

```python
Orders_clean[Orders_clean["Discount %"] > 100]
```

```python
Orders_clean.to_csv("Orders_Clean.csv", index=False)
```

## Finding Maximum Shipping Time 

```python
Max_shipping_time=Orders_clean['Shipping Time'].max()
Max_shipping_time
```

## Average profit Margin

```python
Avg_Profit_Margin=Orders_clean['Profit Margin %'].mean()
Avg_Profit_Margin
```
