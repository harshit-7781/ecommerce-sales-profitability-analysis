# Retail Profitability Intelligence

## Business Problem

ABC Retail experienced a 20% increase in sales, but profit growth remained stagnant.

The objective of this analysis is to identify the factors responsible for declining profitability and recommend actionable business improvements using data.

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd

```

```python
pd.options.display.float_format = '{:,.2f}'.format
```

```python
Orders = pd.read_csv(r'Orders_Clean.csv')
```

```python
Orders.head(3)
```

## Q1) Have sales and profit actually increased over time?

```python
growth=Orders.groupby('Order Year')[['Sales','Profit']].sum()
growth
```

```python
plt.figure(figsize=(6,3))
sns.lineplot(x='Order Year',y='Sales',data=growth)
plt.title("Year-Sales Graph")
plt.show()
```

### Sales has increased over the Years 

```python
plt.figure(figsize=(6,3))
sns.lineplot(x='Order Year',y='Profit',data=growth)
plt.title("Year-Profit Graph")
plt.show
```

### Profit has also increased over the years  

## Q2) Which categories make the most money?

```python
Orders.groupby('Category')[['Profit','Sales','Profit Margin %']].sum().sort_values(by='Profit',ascending=False)
```

## Q3) Which Sub-Categories are hurting the business?

```python
Orders.groupby('Sub-Category')[['Profit','Sales','Profit Margin %']].sum().sort_values(by='Profit Margin %',ascending=True)
```

## Q4) Do discounts reduce profitability?

```python
Orders.groupby('Category')[['Discount','Profit']].sum().sort_values(by = 'Discount', ascending=False)
```

```python
Orders[['Discount','Profit']].corr()
```

```python
Orders.groupby("Category")[["Discount", "Profit"]].mean()
```

```python
Orders.groupby("Discount")["Profit"].mean().sort_index()
```

```python
sns.scatterplot(x='Discount',y='Profit',data=Orders)
plt.show()
```

#### Yes the discounts are reducing the Profitability . 

## Q5) Which states are losing money?

```python
l_states=Orders.groupby('State/Province')[['Sales','Discount','Profit','Profit Margin %']].sum().sort_values(by = 'Profit', ascending=False)
l_states.head(10)
```

```python
l_states.tail(10)
```

```python
state_profit = Orders.groupby("State/Province")["Profit"].sum().reset_index().sort_values( by = 'Profit' ,ascending=True)
state_profit
```

```python
loss_states = state_profit[state_profit["Profit"] < 0]
loss_states
```

```python

sns.barplot(x='Profit',y='State/Province',data=loss_states)
plt.show()
```

## Q6) Which customers generate the most revenue?

```python
Top_customers=Orders.groupby('Customer ID')[['Customer Name','Sales','Profit','Profit Margin %']].sum().sort_values(by = 'Profit', ascending= False)
Top_customers.head(10)
```

## Q7) Which products should management promote?

```python
Promoting_products=Orders.groupby('Sub-Category')[['Sales','Profit','Profit Margin %']].sum().sort_values(by='Profit',ascending=True)
Promoting_products.head(5)
```

## Q8) Which products should be discontinued?

### The products which are making great sales but , are making no profit or negative profit , should be discontinued.


```python
Promoting_products=Orders.groupby('Sub-Category')[['Sales','Profit','Profit Margin %']].sum().sort_values(by='Profit',ascending=True)
Promoting_products.head(3)
```

## Q9) Which shipping mode performs best?

```python
Orders['Ship Mode'].unique()
```

```python
no_of_orders=Orders.groupby('Ship Mode')['Order ID'].count().sort_values(ascending=False)
no_of_orders
```

```python
Orders.groupby('Ship Mode')['Profit'].sum().sort_values(ascending=False)
```

```python
Orders.groupby('Ship Mode')['Sales'].sum().sort_values(ascending=False)
```

```python
Orders[['Order Date', 'Ship Date']].dtypes
```

```python
Orders["Order Date"] = pd.to_datetime(Orders["Order Date"])
Orders["Ship Date"] = pd.to_datetime(Orders["Ship Date"])
Orders[['Order Date', 'Ship Date']].dtypes
```

```python
Orders['Delivery Days']=(Orders['Ship Date'] - Orders['Order Date']).dt.days
Orders.groupby("Ship Mode")["Delivery Days"].mean().sort_values()
```

## Q10) Which months perform best?

```python
monthly_sales=Orders.groupby('Month Name')[['Sales','Profit','Profit Margin %']].sum().sort_values(by='Profit',ascending=False)
monthly_sales
```
